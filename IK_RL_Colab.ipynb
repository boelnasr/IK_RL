## View Results

After training completes, view the generated plots and metrics.

```python
# List output files
import os
from glob import glob

# Find the latest results directory
result_dirs = sorted(glob('results/*'), key=os.path.getmtime, reverse=True)
if result_dirs:
    latest_dir = result_dirs[0]
    print(f"Latest results: {latest_dir}")
    print("\nFiles:")
    for f in os.listdir(latest_dir):
        print(f"  - {f}")
else:
    print("No results found yet.")
```

```python
# Display training plots
from IPython.display import Image, display
import os
from glob import glob

result_dirs = sorted(glob('results/*'), key=os.path.getmtime, reverse=True)
if result_dirs:
    latest_dir = result_dirs[0]
    png_files = glob(os.path.join(latest_dir, '*.png'))
    
    for png in sorted(png_files):
        print(f"\n{os.path.basename(png)}:")
        display(Image(filename=png, width=800))
```

```python
# Display training summary
from glob import glob
import os

result_dirs = sorted(glob('results/*'), key=os.path.getmtime, reverse=True)
if result_dirs:
    latest_dir = result_dirs[0]
    summary_file = os.path.join(latest_dir, 'training_summary.txt')
    if os.path.exists(summary_file):
        with open(summary_file, 'r') as f:
            print(f.read())
```
